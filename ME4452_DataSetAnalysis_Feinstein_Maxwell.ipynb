# Roller Coaster Dataset Exploration

```python
import pandas as pd
import numpy as np
```

#### Read CSV and visually inspect data

```python
coasters_data = "rfdb_rcdb_matched.csv"
coasters_df = pd.read_csv(coasters_data)
coasters_df.head()
```

#### Inspect rows and columns (many columns are not useful to us)

```python
print("Number of Rows: ", coasters_df.shape)
print("Names of Columns: ")
coasters_df.columns
```

#### Drop unnecessary columns (preliminary), rename columns to be more useful, drop units and convert to float64 from str

```python
coasters_df_clean = coasters_df.drop(columns=['ride_name','park','valid_recording_count','miny_iqr','maxy_iqr','minx_iqr','maxx_iqr','minz_iqr',
                                              'maxz_iqr','minc_iqr','maxc_iqr','floater_median', 'floater_iqr', 'flojector_median','flojector_iqr', 
                                              'ejector_median', 'ejector_iqr','rcdb_url','search_url','match_score','rcdb_match_status'])

coasters_df_clean.rename(columns={'rcdb_name':'ride_name','rcdb_park':'park','rcdb_length':'length_ft','rcdb_height':'height_ft','rcdb_speed':'speed_mph',
                                  'rcdb_inversions':'inversions','rcdb_duration':'duration_s','rcdb_elements':'elements','rcdb_drop':'drop_ft',
                                  'rcdb_vertical_angle':'vertical_angle_deg','miny_median':'minz_median_g','maxy_median':'maxz_median_g',
                                  'minx_median':'minx_median_g','maxx_median':'maxx_median_g','minz_median':'miny_median_g','maxz_median':'maxy_median_g',
                                  'minc_median':'minc_median_g','maxc_median':'maxc_median_g'}, inplace=True)

coasters_df_clean = coasters_df_clean[['ride_name','park','manufacturer','length_ft','height_ft','speed_mph','inversions','duration_s','elements','drop_ft',
                                       'vertical_angle_deg','minz_median_g','maxz_median_g','minx_median_g','maxx_median_g','miny_median_g','maxy_median_g',
                                       'minc_median_g','maxc_median_g']]

coasters_df_clean['length_ft'] = coasters_df_clean['length_ft'].str.replace('ft', '', case=False)
coasters_df_clean['length_ft'] = pd.to_numeric(coasters_df_clean['length_ft'])

coasters_df_clean['height_ft'] = coasters_df_clean['height_ft'].str.replace('ft', '', case=False)
coasters_df_clean['height_ft'] = pd.to_numeric(coasters_df_clean['height_ft'])

coasters_df_clean['speed_mph'] = coasters_df_clean['speed_mph'].str.replace('mph', '', case=False)
coasters_df_clean['speed_mph'] = pd.to_numeric(coasters_df_clean['speed_mph'])

coasters_df_clean['drop_ft'] = coasters_df_clean['drop_ft'].str.replace('ft', '', case=False)
coasters_df_clean['drop_ft'] = pd.to_numeric(coasters_df_clean['drop_ft'])

coasters_df_clean['vertical_angle_deg'] = coasters_df_clean['vertical_angle_deg'].str.replace('°', '', case=False)
coasters_df_clean['vertical_angle_deg'] = pd.to_numeric(coasters_df_clean['vertical_angle_deg'])

coasters_df_clean['duration_s'] = pd.to_timedelta('00:' + coasters_df_clean['duration_s']).dt.total_seconds()

coasters_df_clean.head()
```

#### Inspect non-null count in each column to evaluate which rows to drop

```python
coasters_df_clean.info()
```

#### Drop rows with empty entries for length, height, and speed (these inputs are key and must be present)

```python
coasters_df_clean = coasters_df_clean.dropna(subset=['length_ft','height_ft','speed_mph'])
coasters_df_clean.info()
```

#### Drop partially filled columns

```python
coasters_df_clean = coasters_df_clean.drop(columns=['duration_s','elements','drop_ft','vertical_angle_deg'])
coasters_df_clean.info()
```

```python
coasters_df_clean.describe()
```

#### Save cleaned dataframe to CSV and open as new dataframe

```python
coasters_df_clean.to_csv("coasters_dataset.csv", index=False)
coasters_df = pd.read_csv("coasters_dataset.csv")
```

#### Plot length, height, speed, and number of inversions vs. max and min lateral, longitudinal, and vertical g-forces

```python
import matplotlib.pyplot as plt
```

```python
g_force_columns = ['miny_median_g', 'maxy_median_g', 
                   'minx_median_g', 'maxx_median_g', 
                   'minz_median_g', 'maxz_median_g', 
                   'minc_median_g', 'maxc_median_g']

g_force_labels_min = {'Lateral': 'miny_median_g', 
                      'Longitudinal': 'minx_median_g', 
                      'Vertical': 'minz_median_g', 
                      'Combined': 'minc_median_g'}

g_force_labels_max =  {'Lateral': 'maxy_median_g', 
                       'Longitudinal': 'maxx_median_g',  
                       'Vertical': 'maxz_median_g', 
                       'Combined': 'maxc_median_g'}

ride_measurements = {'length_ft':'Length [ft]', 'height_ft':'Height [ft]', 'speed_mph':'Speed [mph]','inversions':'Number of Inversions'}
```

```python
for measurement, x_label in ride_measurements.items():
    
    plt.figure(figsize=(11, 7))
    for g_label, g_force in g_force_labels_min.items():
        plt.scatter(coasters_df[[measurement]], coasters_df[[g_force]], label=g_label, s=10, alpha=0.7)
    plt.xlabel(x_label)
    plt.ylabel('Median Minimum G-Force [g]')
    plt.title(f'{x_label} vs. Minimum G-Forces')
    plt.legend(title='G-force Direction',loc='best')
    plt.grid(alpha=0.3)
    plt.show()

    plt.figure(figsize=(11, 7))
    for g_label, g_force in g_force_labels_max.items():
        plt.scatter(coasters_df[[measurement]], coasters_df[[g_force]], label=g_label, s=10, alpha=0.7)
    plt.xlabel(x_label)
    plt.ylabel('Median Maximum G-Force [g]')
    plt.title(f'{x_label} vs. Maximum G-Forces')
    plt.legend(title='G-force Direction',loc='best')
    plt.grid(alpha=0.3)
    plt.show()
```
