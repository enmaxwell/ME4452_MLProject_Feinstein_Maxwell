```python
import pandas as pd
import numpy as np
```

```python
coasters_data = "rfdb_rcdb_matched.csv"
coasters_df = pd.read_csv(coasters_data)
coasters_df.head()
```

```python
print("Number of Rows: ", coasters_df.shape)
print("Number of Columns: ")
coasters_df.columns
```

```python
coasters_df_clean = coasters_df.drop(columns=['ride_name','park','valid_recording_count','miny_iqr','maxy_iqr','minx_iqr','maxx_iqr','minz_iqr',
                                              'maxz_iqr','minc_iqr','maxc_iqr','floater_median', 'floater_iqr', 'flojector_median','flojector_iqr', 
                                              'ejector_median', 'ejector_iqr','rcdb_url','search_url','match_score','rcdb_match_status'])

coasters_df_clean.rename(columns={'rcdb_name':'ride_name','rcdb_park':'park','rcdb_length':'length_ft','rcdb_height':'height_ft','rcdb_speed':'speed_mph',
                                  'rcdb_inversions':'inversions','rcdb_duration':'duration_s','rcdb_elements':'elements','rcdb_drop':'drop_ft',
                                  'rcdb_vertical_angle':'vertical_angle_deg'}, inplace=True)

coasters_df_clean = coasters_df_clean[['ride_name','park','manufacturer','length_ft','height_ft','speed_mph','inversions','duration_s','elements','drop_ft',
                                       'vertical_angle_deg','miny_median','maxy_median','minx_median','maxx_median','minz_median','maxz_median',
                                       'minc_median','maxc_median']]

coasters_df_clean['length_ft'] = coasters_df_clean['length_ft'].str.replace('ft', '', case=False)
coasters_df_clean['length_ft'] = pd.to_numeric(coasters_df_clean['length_ft'])

coasters_df_clean['height_ft'] = coasters_df_clean['height_ft'].str.replace('ft', '', case=False)
coasters_df_clean['height_ft'] = pd.to_numeric(coasters_df_clean['height_ft'])

coasters_df_clean['speed_mph'] = coasters_df_clean['speed_mph'].str.replace('mph', '', case=False)
coasters_df_clean['speed_mph'] = pd.to_numeric(coasters_df_clean['speed_mph'])

coasters_df_clean['drop_ft'] = coasters_df_clean['drop_ft'].str.replace('ft', '', case=False)
coasters_df_clean['drop_ft'] = pd.to_numeric(coasters_df_clean['drop_ft'])

coasters_df_clean['vertical_angle_deg'] = coasters_df_clean['vertical_angle_deg'].str.replace('°', '', case=False)
coasters_df_clean['vertical_angle_deg'] = pd.to_numeric(coasters_df_clean['vertical_angle_deg'])

coasters_df_clean['duration_s'] = pd.to_timedelta('00:' + coasters_df_clean['duration_s']).dt.total_seconds()

coasters_df_clean.head()
```
